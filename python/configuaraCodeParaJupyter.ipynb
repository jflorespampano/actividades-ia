## Crear entorno virtual (kernel)

Si desea crear y registrar en jupyter un kernel de python3.12:

1. Desacargue python 3.12 si no lo tiene [desde](https://www.python.org/downloads/windows/)
    * para saber si lo tiene o no, en una ventana de bash ponga `where python`
2. Ya que lo tenga, para crear y registrar el entorno:

```bash
mkdir mientorno
cd mientorno

py -3.12 -m venv mientorno
source mientorno/Scripts/avtivate
pyhton --version

# instalar biblotecas si es necesario
pip install numpy pandas matplotlib tensorflow

# para registrar el kernel en jupyter
pip install ipykernel
# registra el kernel
python -m ipykernel install --user --name=mientorno --display-name="kernel(pyhton 3.12)"

# cerrar el entorno
deactivate
```

Listo, ahora puede seleccionar este kernel es sus cuadernos desde Code.
