# xor

Clasificar las salidas del xor

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
# 1. Datos de entrada (XOR)
# Entradas
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
], dtype=np.float32)

# Salidas esperadas (XOR)
y = np.array([
    [0],
    [1],
    [1],
    [0]
], dtype=np.float32)
```

```python
# 2. Construcción del modelo
model = keras.Sequential([
    layers.Input(shape=(2,)),           # 2 entradas
    layers.Dense(8, activation='relu'), # Capa oculta
    layers.Dense(4, activation='relu'), # Segunda capa oculta (opcional)
    layers.Dense(1, activation='sigmoid') # Salida binaria
])
```

```python
# 3. Compilación
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
# 4. Entrenamiento
history = model.fit(
    X, y,
    epochs=1000,
    verbose=0  # Cambia a 1 para ver el progreso
)
```

```python
# 5. Evaluación
loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"\nLoss: {loss:.4f} | Accuracy: {accuracy:.4f}")

# 6. Predicciones
print("\nPredicciones:")
predicciones = model.predict(X, verbose=0)

for i in range(len(X)):
    entrada = X[i]
    real = y[i][0]
    prob = predicciones[i][0]
    clase = 1 if prob > 0.5 else 0
    print(f"Entrada: {entrada} -> Real: {int(real)} | "
          f"Prob: {prob:.4f} | Predicción: {clase}")
```
